```python

# Fase 4 - Data Analytics
# Proyecto: Appliances Energy Prediction
# Paso 1: Carga inicial del conjunto de datos

import pandas as pd

# Leer el archivo CSV que ya está cargado en Colab
df = pd.read_csv('/content/appliances+energy+prediction.csv')

# Mostrar las primeras cinco filas
display(df.head())

# Confirmar que el archivo se cargó correctamente
print("Dataset cargado correctamente")
print("Dimensiones:", df.shape)
```

```python

# =====================================================
# PASO 2: EXPLORACIÓN DE LA ESTRUCTURA DEL DATASET
# Proyecto: Appliances Energy Prediction
# =====================================================

# 1. Número de registros y columnas
print("1. DIMENSIONES DEL DATASET")
print("Número de registros:", df.shape[0])
print("Número de columnas:", df.shape[1])

# 2. Nombres de las variables
print("\n2. VARIABLES DISPONIBLES")
for numero, columna in enumerate(df.columns, start=1):
    print(f"{numero}. {columna}")

# 3. Información general del dataset
print("\n3. INFORMACIÓN GENERAL")
df.info()

# 4. Verificación de valores faltantes
print("\n4. VALORES FALTANTES POR COLUMNA")
display(df.isnull().sum().to_frame(name="Valores faltantes"))

# 5. Total de valores faltantes
print("\nTotal de valores faltantes:", df.isnull().sum().sum())

# 6. Estadísticas descriptivas
print("\n5. ESTADÍSTICAS DESCRIPTIVAS")
display(df.describe())

# 7. Primeras cinco filas del dataset
print("\n6. PRIMEROS CINCO REGISTROS")
display(df.head())
```

```python

# =====================================================
# PASO 3: PREPARACIÓN Y VALIDACIÓN DE LOS DATOS
# Proyecto: Appliances Energy Prediction
# =====================================================

# 1. Convertir la variable date a formato datetime
df['date'] = pd.to_datetime(df['date'])

print("1. CONVERSIÓN DE LA VARIABLE DATE")
print("Tipo de dato:", df['date'].dtype)

# 2. Verificar registros duplicados
print("\n2. VERIFICACIÓN DE REGISTROS DUPLICADOS")

duplicados = df.duplicated().sum()

print("Número de registros duplicados:", duplicados)

# 3. Verificar nuevamente las dimensiones
print("\n3. DIMENSIONES ACTUALES DEL DATASET")
print("Registros:", df.shape[0])
print("Columnas:", df.shape[1])

# 4. Verificar el rango de fechas
print("\n4. RANGO DE FECHAS")
print("Fecha inicial:", df['date'].min())
print("Fecha final:", df['date'].max())

# 5. Verificar los tipos de datos actualizados
print("\n5. TIPOS DE DATOS ACTUALIZADOS")
print(df.dtypes)
```

```python

# =====================================================
# PASO 4: ANÁLISIS DESCRIPTIVO Y VISUALIZACIÓN
# Proyecto: Appliances Energy Prediction
# =====================================================

# Importar librerías para visualización
import matplotlib.pyplot as plt
import seaborn as sns

# Configuración de gráficos
sns.set_theme(style="whitegrid")

# 1. Histograma del consumo energético
plt.figure(figsize=(10, 5))

sns.histplot(
    data=df,
    x='Appliances',
    bins=30,
    kde=True
)

plt.title('Distribution of Household Appliance Energy Consumption')
plt.xlabel('Appliances Energy Consumption (Wh)')
plt.ylabel('Frequency')

plt.tight_layout()
plt.show()
```

```python

# =====================================================
# PASO 5: IDENTIFICACIÓN DE VALORES ATÍPICOS
# Proyecto: Appliances Energy Prediction
# =====================================================

# Crear figura
plt.figure(figsize=(10, 5))

# Generar diagrama de caja
sns.boxplot(
    x=df['Appliances'],
    color='skyblue'
)

# Configurar título y ejes
plt.title('Boxplot of Household Appliance Energy Consumption')
plt.xlabel('Appliances Energy Consumption (Wh)')

# Mostrar gráfico
plt.tight_layout()
plt.show()
```

```python

# =====================================================
# PASO 6: RELACIÓN ENTRE TEMPERATURA Y CONSUMO
# Proyecto: Appliances Energy Prediction
# =====================================================

plt.figure(figsize=(10, 5))

# Gráfico de dispersión
sns.scatterplot(
    data=df,
    x='T1',
    y='Appliances',
    alpha=0.4
)

# Títulos y etiquetas
plt.title('Relationship Between Indoor Temperature and Appliance Energy Consumption')
plt.xlabel('Indoor Temperature T1 (°C)')
plt.ylabel('Appliance Energy Consumption (Wh)')

plt.tight_layout()
plt.show()
```

```python

# =====================================================
# PASO 7: ANÁLISIS DE CORRELACIÓN
# Proyecto: Appliances Energy Prediction
# =====================================================

# Seleccionar las variables
temperatura = df['T1']
consumo = df['Appliances']

# Calcular correlación de Pearson
pearson = temperatura.corr(consumo, method='pearson')

# Calcular correlación de Spearman
spearman = temperatura.corr(consumo, method='spearman')

# Mostrar resultados
print("CORRELACIÓN ENTRE TEMPERATURA Y CONSUMO")
print("----------------------------------------")

print(f"Correlación de Pearson: {pearson:.4f}")
print(f"Correlación de Spearman: {spearman:.4f}")

# Interpretación orientativa de Pearson
if abs(pearson) < 0.10:
    print("\nInterpretación: correlación lineal prácticamente nula.")
elif abs(pearson) < 0.30:
    print("\nInterpretación: correlación lineal débil.")
elif abs(pearson) < 0.50:
    print("\nInterpretación: correlación lineal moderada.")
elif abs(pearson) < 0.70:
    print("\nInterpretación: correlación lineal considerable.")
else:
    print("\nInterpretación: correlación lineal fuerte.")
```

```python

# =====================================================
# PASO 8: MATRIZ DE CORRELACIÓN
# Proyecto: Appliances Energy Prediction
# =====================================================

# Seleccionar variables ambientales y consumo
variables = [
    'Appliances',
    'T1',
    'RH_1',
    'T_out',
    'RH_out',
    'Windspeed',
    'Visibility',
    'Tdewpoint'
]

# Crear matriz de correlación de Pearson
matriz_corr = df[variables].corr(method='pearson')

# Mostrar matriz numérica
print("MATRIZ DE CORRELACIÓN DE PEARSON")
display(matriz_corr.round(3))

# Generar mapa de calor
plt.figure(figsize=(11, 8))

sns.heatmap(
    matriz_corr,
    annot=True,
    cmap='coolwarm',
    fmt='.2f',
    linewidths=0.5,
    center=0
)

plt.title('Correlation Matrix: Energy Consumption and Environmental Variables')
plt.tight_layout()
plt.show()
```

```python

# =====================================================
# PASO 9: CORRELACIÓN DE TODAS LAS VARIABLES
# Proyecto: Appliances Energy Prediction
# =====================================================

# Seleccionar todas las variables numéricas
df_numerico = df.select_dtypes(include=['number'])

# Calcular correlación de Pearson con Appliances
correlaciones = (
    df_numerico.corr()['Appliances']
    .drop('Appliances')
    .sort_values(ascending=False)
)

# Mostrar resultados ordenados
print("CORRELACIÓN DE TODAS LAS VARIABLES CON APPLIANCES")
display(correlaciones.to_frame(name='Correlación de Pearson').round(4))

# Crear gráfico de barras
plt.figure(figsize=(12, 9))

colores = [
    'steelblue' if valor >= 0 else 'indianred'
    for valor in correlaciones.values
]

correlaciones.sort_values().plot(
    kind='barh',
    color=colores
)

plt.title('Correlation of All Numerical Variables with Appliance Energy Consumption')
plt.xlabel('Pearson Correlation Coefficient')
plt.ylabel('Variables')
plt.axvline(x=0, color='black', linewidth=0.8)

plt.tight_layout()
plt.show()
```

```python

# PASO 9.1: AJUSTE DE LA VISUALIZACIÓN

# Ordenar las correlaciones
correlaciones_ordenadas = correlaciones.sort_values()

# Asignar colores según el signo correcto
colores = [
    'steelblue' if valor >= 0 else 'indianred'
    for valor in correlaciones_ordenadas.values
]

# Crear gráfico
plt.figure(figsize=(12, 9))

plt.barh(
    correlaciones_ordenadas.index,
    correlaciones_ordenadas.values,
    color=colores
)

plt.title('Correlation of All Numerical Variables with Appliance Energy Consumption')
plt.xlabel('Pearson Correlation Coefficient')
plt.ylabel('Variables')
plt.axvline(x=0, color='black', linewidth=0.8)

plt.tight_layout()
plt.show()
```

```python

# =====================================================
# PASO 10: ANÁLISIS TEMPORAL DEL CONSUMO ENERGÉTICO
# Proyecto: Appliances Energy Prediction
# =====================================================

# Crear gráfico de consumo a través del tiempo
plt.figure(figsize=(14, 5))

plt.plot(
    df['date'],
    df['Appliances'],
    linewidth=0.6,
    alpha=0.7
)

plt.title('Household Appliance Energy Consumption Over Time')
plt.xlabel('Date')
plt.ylabel('Energy Consumption (Wh)')

plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python

# =====================================================
# PASO 11: ANÁLISIS DEL CONSUMO PROMEDIO DIARIO
# Proyecto: Appliances Energy Prediction
# =====================================================

# Crear una serie temporal con el promedio diario
consumo_diario = (
    df.set_index('date')['Appliances']
    .resample('D')
    .mean()
)

# Mostrar estadísticas del consumo diario
print("ESTADÍSTICAS DEL CONSUMO PROMEDIO DIARIO")
print("Promedio diario:", round(consumo_diario.mean(), 2), "Wh")
print("Consumo diario máximo:", round(consumo_diario.max(), 2), "Wh")
print("Consumo diario mínimo:", round(consumo_diario.min(), 2), "Wh")

# Crear gráfico
plt.figure(figsize=(14, 5))

plt.plot(
    consumo_diario.index,
    consumo_diario.values,
    marker='o',
    markersize=3,
    linewidth=1.2
)

plt.title('Daily Average Household Appliance Energy Consumption')
plt.xlabel('Date')
plt.ylabel('Average Energy Consumption (Wh)')

plt.xticks(rotation=45)
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```

```python

# =====================================================
# PASO 12: CONSUMO ENERGÉTICO POR DÍA DE LA SEMANA
# Proyecto: Appliances Energy Prediction
# =====================================================

# Crear columna con el día de la semana
df['DayOfWeek'] = df['date'].dt.day_name()

# Orden correcto de los días
orden_dias = [
    'Monday',
    'Tuesday',
    'Wednesday',
    'Thursday',
    'Friday',
    'Saturday',
    'Sunday'
]

# Calcular consumo promedio por día
consumo_semana = (
    df.groupby('DayOfWeek')['Appliances']
    .mean()
    .reindex(orden_dias)
)

# Mostrar resultados
print("CONSUMO PROMEDIO POR DÍA DE LA SEMANA")
display(consumo_semana.round(2).to_frame(
    name='Consumo promedio (Wh)'
))

# Crear gráfico de barras
plt.figure(figsize=(10, 5))

sns.barplot(
    x=consumo_semana.index,
    y=consumo_semana.values
)

plt.title('Average Appliance Energy Consumption by Day of Week')
plt.xlabel('Day of Week')
plt.ylabel('Average Energy Consumption (Wh)')

plt.xticks(rotation=30)
plt.tight_layout()
plt.show()
```

```python

# =====================================================
# PASO 13: COMPARACIÓN ENTRE DÍAS LABORALES Y FINES
# DE SEMANA
# Proyecto: Appliances Energy Prediction
# =====================================================

import numpy as np

# Clasificar los días
df['DayType'] = np.where(
    df['date'].dt.dayofweek >= 5,
    'Weekend',
    'Weekday'
)

# Calcular estadísticas por grupo
comparacion = df.groupby('DayType')['Appliances'].agg(
    ['mean', 'median', 'std', 'count']
)

print("COMPARACIÓN DEL CONSUMO ENERGÉTICO")
display(comparacion.round(2))

# Calcular diferencia porcentual
laboral = comparacion.loc['Weekday', 'mean']
fin_semana = comparacion.loc['Weekend', 'mean']

diferencia = fin_semana - laboral
porcentaje = (diferencia / laboral) * 100

print("\nRESULTADOS DE LA COMPARACIÓN")
print(f"Promedio días laborales: {laboral:.2f} Wh")
print(f"Promedio fines de semana: {fin_semana:.2f} Wh")
print(f"Diferencia: {diferencia:.2f} Wh")
print(f"Diferencia porcentual: {porcentaje:.2f}%")

# Crear gráfico comparativo
plt.figure(figsize=(8, 5))

sns.barplot(
    data=df,
    x='DayType',
    y='Appliances',
    errorbar=None
)

plt.title('Average Appliance Energy Consumption: Weekdays vs Weekends')
plt.xlabel('Day Type')
plt.ylabel('Average Energy Consumption (Wh)')

plt.tight_layout()
plt.show()
```

```python

# =====================================================
# PASO 14: PRUEBA ESTADÍSTICA MANN-WHITNEY
# Proyecto: Appliances Energy Prediction
# =====================================================

from scipy.stats import mannwhitneyu

# Calcular consumo promedio por día
consumo_diario = (
    df.set_index('date')['Appliances']
    .resample('D')
    .mean()
    .dropna()
)

# Clasificar cada día
dias_laborales = consumo_diario[
    consumo_diario.index.dayofweek < 5
]

fines_semana = consumo_diario[
    consumo_diario.index.dayofweek >= 5
]

# Aplicar prueba Mann-Whitney
estadistico, p_valor = mannwhitneyu(
    dias_laborales,
    fines_semana,
    alternative='two-sided'
)

# Mostrar resultados
print("PRUEBA ESTADÍSTICA MANN-WHITNEY")
print("--------------------------------")

print("Días laborales:", len(dias_laborales))
print("Fines de semana:", len(fines_semana))

print(f"Estadístico U: {estadistico:.2f}")
print(f"Valor p: {p_valor:.6f}")

# Interpretación
nivel_significancia = 0.05

if p_valor < nivel_significancia:
    print("\nResultado: Se encontró una diferencia estadísticamente significativa.")
else:
    print("\nResultado: No se encontró evidencia suficiente de una diferencia estadísticamente significativa.")
```
